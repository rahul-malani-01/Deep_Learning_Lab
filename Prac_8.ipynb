```python
# ============================================================
# Rahul malani - 23070521088
# STOCK MARKET PREDICTION USING LSTM
# Dataset: nasdq.csv
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout


# ============================================================
# 2. LOAD THE DATASET
# ============================================================

file_path = "/mnt/data/nasdq.csv"

data = pd.read_csv("/content/nasdq.csv")

print("Dataset loaded successfully!")

print("\nFirst 5 rows:")
print(data.head())

print("\nDataset shape:")
print(data.shape)

print("\nColumns:")
print(data.columns.tolist())


# ============================================================
# 3. CHECK MISSING VALUES
# ============================================================

print("\nMissing values:")
print(data.isnull().sum())


# ============================================================
# 4. CONVERT DATE COLUMN
# ============================================================

data["Date"] = pd.to_datetime(data["Date"])

# Sort according to date
data = data.sort_values("Date")

# Reset index
data = data.reset_index(drop=True)


# ============================================================
# 5. SELECT CLOSE PRICE
# ============================================================

# We will use Close price as the target variable

close_data = data[["Date", "Close"]].copy()

# Remove missing Close values
close_data = close_data.dropna()

print("\nClose price data:")
print(close_data.head())

print("\nNumber of records:", len(close_data))


# ============================================================
# 6. VISUALIZE ORIGINAL STOCK PRICE
# ============================================================

plt.figure(figsize=(14, 6))

plt.plot(
    close_data["Date"],
    close_data["Close"]
)

plt.title("NASDAQ Historical Stock Closing Price")

plt.xlabel("Date")
plt.ylabel("Closing Price")

plt.grid(True)

plt.show()


# ============================================================
# 7. EXTRACT CLOSE VALUES
# ============================================================

prices = close_data["Close"].values.reshape(-1, 1)



# ============================================================
# 8. NORMALIZE THE DATA
# ============================================================

scaler = MinMaxScaler(feature_range=(0, 1))

scaled_prices = scaler.fit_transform(prices)

print("\nScaled data shape:")
print(scaled_prices.shape)


# ============================================================
# 9. TRAIN-TEST SPLIT
# ============================================================

# 80% training
# 20% testing

training_size = int(len(scaled_prices) * 0.80)

train_data = scaled_prices[:training_size]

test_data = scaled_prices[training_size:]


print("\nTraining data size:", len(train_data))

print("Testing data size:", len(test_data))

















# Rahul Malani - 23070521088
# ============================================================
# 10. CREATE SEQUENTIAL TRAINING DATA
# ============================================================

# Use previous 60 days
# to predict the next day

time_step = 60
X_train = []
y_train = []

for i in range(time_step, len(train_data)):
    # Previous 60 days
    X_train.append(
        train_data[i-time_step:i, 0]
    )
    # Next day's price
    y_train.append(
        train_data[i, 0]
    )

# Convert to NumPy arrays
X_train = np.array(X_train)
y_train = np.array(y_train)

print("\nTraining sequences created!")
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)

# ============================================================
# 11. CREATE TESTING SEQUENCES
# ============================================================

# We need the last 60 training values
# to make the first testing prediction.

combined_data = np.concatenate(
    (train_data[-time_step:], test_data),
    axis=0
)

X_test = []
y_test = []

for i in range(time_step, len(combined_data)):
    # Previous 60 days
    X_test.append(
        combined_data[i-time_step:i, 0]
    )
    # Actual next day
    y_test.append(
        combined_data[i, 0]
    )

X_test = np.array(X_test)
y_test = np.array(y_test)

print("\nTesting sequences created!")
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)

# ============================================================
# 12. RESHAPE DATA FOR LSTM
# ============================================================

# LSTM expects 3D data:
# (samples, time_steps, features)

X_train = X_train.reshape(
    X_train.shape[0],
    X_train.shape[1],
    1
)

X_test = X_test.reshape(
    X_test.shape[0],
    X_test.shape[1],
    1
)

print("\nFinal LSTM input shapes:")
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

# ============================================================
# 13. BUILD LSTM MODEL
# ============================================================

model = Sequential()

# First LSTM layer

model.add(
    LSTM(
        50,
        return_sequences=True,
        input_shape=(time_step, 1)
    )
)
# Dropout

model.add(
    Dropout(0.2)
)

# Second LSTM layer

model.add(
    LSTM(
        50,
        return_sequences=False
    )
)

# Dropout

model.add(
    Dropout(0.2)
)


# Output layer
model.add(
    Dense(1)
)

# ============================================================
# 14. COMPILE MODEL
# ============================================================

model.compile(
    optimizer="adam",
    loss="mean_squared_error"
)

print("\n==============================")
print("LSTM MODEL")
print("==============================")

model.summary()

# ============================================================
# 15. TRAIN THE MODEL
# ============================================================

history = model.fit(
    X_train,
    y_train,
    epochs=20,
    batch_size=32,
    validation_split=0.1,
    verbose=1
)

# ============================================================
# 16. MAKE PREDICTIONS
# ============================================================

predicted_scaled = model.predict(X_test)

# ============================================================
# 17. CONVERT PREDICTIONS BACK TO ORIGINAL PRICE
# ============================================================

predicted_prices = scaler.inverse_transform(
    predicted_scaled
)

actual_prices = scaler.inverse_transform(
    y_test.reshape(-1, 1)
)

# ============================================================
# 18. CALCULATE PERFORMANCE METRICS
# ============================================================

mae = mean_absolute_error(
    actual_prices,
    predicted_prices
)

mse = mean_squared_error(
    actual_prices,
    predicted_prices
)

rmse = np.sqrt(mse)

print("\n====================================")
print("MODEL PERFORMANCE")
print("====================================")
print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
print("Root Mean Squared Error (RMSE):", rmse)

# ============================================================
# 19. PLOT TRAINING AND VALIDATION LOSS
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("LSTM Training vs Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()

# ============================================================
# 20. CREATE TEST DATES
# ============================================================

test_dates = close_data["Date"].iloc[
    training_size:
].reset_index(drop=True)






# ============================================================
# 21. ACTUAL VS PREDICTED STOCK PRICE
# ============================================================

plt.figure(figsize=(14, 6))

plt.plot(
    test_dates,
    actual_prices.flatten(),
    label="Actual Price"
)

plt.plot(
    test_dates,
    predicted_prices.flatten(),
    label="Predicted Price"
)

plt.title(
    "Actual vs Predicted NASDAQ Stock Prices"
)


plt.xlabel("Date")
plt.ylabel("Stock Price")
plt.legend()
plt.grid(True)
plt.show()

# ============================================================
# 22. CREATE RESULTS TABLE
# ============================================================

results = pd.DataFrame({

    "Date": test_dates,

    "Actual Price":
        actual_prices.flatten(),

    "Predicted Price":
        predicted_prices.flatten()

})


print("\n====================================")
print("ACTUAL VS PREDICTED VALUES")
print("====================================")

print(results.head(20))
```
